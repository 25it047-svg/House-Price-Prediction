```python
import pandas as pd

data = {
    "area": [800, 1000, 1200, 1500, 1800, 2000, 2200, 2500],
    "bedrooms": [1, 2, 2, 3, 3, 4, 4, 5],
    "price": [25, 35, 40, 55, 65, 75, 85, 100]
}

df = pd.DataFrame(data)

print(df)
```

```python
from sklearn.linear_model import LinearRegression

X = df[["area", "bedrooms"]]
y = df["price"]

model = LinearRegression()

model.fit(X, y)

print("House Price Prediction Model Trained Successfully")
```

```python
house = [[1600, 3]]

prediction = model.predict(house)

print("Predicted House Price:", round(prediction[0], 2), "Lakhs")
```

```python
house = [[2100, 4]]

prediction = model.predict(house)

print("Predicted House Price:", round(prediction[0], 2), "Lakhs")
```

```python
area = int(input("Enter House Area (sq.ft): "))
bedrooms = int(input("Enter Number of Bedrooms: "))

house = [[area, bedrooms]]

prediction = model.predict(house)

print("\nEstimated House Price:", round(prediction[0], 2), "Lakhs")
```
